### Synthetic Data Generation
This notebook generates a structured synthetic dataset representing human behavioral patterns under different scenarios. The data is designed using controlled mappings between scenario type, difficulty level, emotional state, response time, and decision outcome.

Unlike real-world data, this synthetic dataset ensures consistency and interpretability, enabling the behavioral model to learn clear decision patterns. This forms the foundation for training the decision prediction component of the Human Digital Twin (HDT) system.

```python
import pandas as pd
import random
import numpy as np
random.seed(42)
np.random.seed(42)
import joblib
print("Libraries loaded successfully")
```

```python
scenarios = ["career", "money", "ethics", "relationship", "health", "risk", "conflict", "deadline"]
difficulty_map = {"career" : ["medium", "high", "very high"], "money" : ["medium", "high"], "ethics" : ["high", "very high"], "relationship" : ["medium", "high"], "health" : ["high", "very high"], "risk" : ["low", "medium", "high"], "conflict" : ["medium", "high"], "deadline" : ["high", "very high"]}
emotion_map = {"low" : ["calm", "focused"], "medium" : ["concerned", "thinking"], "high" : ["anxious", "stressed"], "very high" : ["panic", "overwhelmed"]}
print("Categories defined successfully")
```

```python
def generate_behavior_row():
    scenario = random.choice(scenarios)
    difficulty = random.choice(difficulty_map[scenario])
    emotional_state = random.choice(emotion_map[difficulty])
    personality = random.choice(["introvert", "extrovert", "balanced"])
    age_group = random.choice(["young", "middle", "senior"])
    stress_level = random.choice(["low", "medium", "high"])
    decision_context = random.choice(["personal", "professional", "social"])
    if emotional_state in ["calm", "focused"]:
        base_decision = "analytical"
    elif emotional_state in ["concerned", "thinking"]:
        base_decision = "cautious"
    elif emotional_state in ["anxious","stressed"]:
        base_decision = "avoidant"
    elif emotional_state in ["panic", "overwhelmed"]:
        base_decision = "impulsive"
    else:
        base_decision = "proactive"
    if stress_level == "high" and base_decision == "analytical":
        base_decision = "cautious"
    if stress_level == "low" and base_decision == "avoidant":
        base_decision = "cautious"
    if random.random() < 0.08:
        base_decision = random.choice(["analytical", "avoidant", "cautious", "impulsive", "proactive"])
    decision = base_decision        
    response_time_map = {"low" : ["fast", "moderate"], "medium" : ["moderate", "slow"], "high" : ["slow", "very slow"], "very high" : ["very slow"]}
    response_time = random.choice(response_time_map[difficulty])
    return [scenario, difficulty, emotional_state, decision, response_time, personality, age_group, stress_level, decision_context]
```

```python
data = [generate_behavior_row() for i in range(5000)]
df = pd.DataFrame(data, columns = ["scenario", "difficulty", "emotion", "decision", "response_time", "personality", "age_group", "stress_level", "context"])
df = df.drop_duplicates()
print("Rows generated", len(df))
print(df.shape)
display(df.head())
```

```python
joblib.dump(scenarios, "../models/scenarios.pkl")
joblib.dump(difficulty_map, "../models/difficulty_map.pkl")
print("Scenarios and difficulty map saved successfully")
df.to_csv("../data/synthetic.csv", index=False)
print("Synthetic dataset created successfully")
```
