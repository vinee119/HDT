### Synthetic Feature Engineering
This notebook focuses on refining and transforming the synthetic dataset into optimized feature representations. It enhances the quality of input data by structuring encoded features and preparing them for machine learning models.

The goal of this stage is to ensure that the behavioral model receives clean, consistent, and well-structured inputs, improving prediction performance and generalization.

```python
import numpy as np
from sklearn.preprocessing import StandardScaler
import joblib
```

```python
features_encoded = np.load("../data/synthetic_features.npy")
target = np.load("../data/synthetic_labels.npy", allow_pickle = True)
print("Loaded features")
print(features_encoded.shape)
print("Loaded target")
print(target.shape)
```

```python
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features_encoded)
print("Scaled features shape")
print(features_scaled.shape)
```

```python
np.save("../data/synthetic_features_scaled.npy", features_scaled)
np.save("../data/synthetic_labels_final.npy", target)
joblib.dump(scaler, "../models/synthetic_scaler.pkl")
print("Feature engineering complete Scaled features, final labels and scaler saved.")
```
