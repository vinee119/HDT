### Data Preprocessing and Feature Preparation
This notebook performs preprocessing and feature preparation for the synthetic dataset. Categorical variables such as scenario type, difficulty level, emotional state, and response time are transformed into machine-readable format using encoding techniques.

Feature scaling is also applied to normalize the data distribution, ensuring stable and efficient model training. The processed features and labels are then saved for use in the behavioral model training phase.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
import numpy as np
import joblib
```

```python
synthetic_path = ("../data/synthetic.csv")
df = pd.read_csv(synthetic_path)
print("\n=== Dataset Shape ===")
print(df.shape)
print("\n=== Dataset Info ===")
display(df.head())
```

```python
features = df.drop(columns = ["decision"])
target = df["decision"]              
print("Features shape")
print(features.shape)
print("Target shape")
print(target.shape)     
```

```python
encoder = OneHotEncoder(sparse_output = False)
features_encoded = encoder.fit_transform(features) 
print("Encoded feature shape")
print(features_encoded.shape)
```

```python
np.save("../data/synthetic_features.npy", features_encoded)
np.save("../data/synthetic_labels.npy", target)
joblib.dump(encoder, "../models/synthetic_encoder.pkl")
print("All Saved Successfully")
```
